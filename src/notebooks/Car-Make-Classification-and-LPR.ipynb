**Classification Module**

```python
##################################
#### classification_module.py ####
##################################


import os
import cv2
import time
import torch
from collections import defaultdict

# ---------------------- Car Detection + Classification ----------------------

def process_video(
    video_path,
    detection_model,
    classification_model,
    output_path,
    save_crops=False,
    root_folder=None,
    DETECTION_CONF_THRESH=0.0,
    CLASSIFICATION_CONF_THRESH=0.0,
):
    cap = cv2.VideoCapture(video_path)

    # Getting video properties
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Defining output video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    detection_model.eval()
    classification_model.eval()

    frame_count = 0
    class_counts = defaultdict(int)

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            print('End of video')
            break

        frame_count += 1
        print(f"Processing frame {frame_count}")
        start_time = time.time()

        detection_results = detection_model(frame)

        for result in detection_results:
            bboxes = result.boxes.xyxy.cpu().numpy()
            scores = result.boxes.conf.cpu().numpy()
            class_ids = result.boxes.cls.cpu().numpy()

            for i, (x1, y1, x2, y2) in enumerate(bboxes):
                if scores[i] < DETECTION_CONF_THRESH:
                    continue
                if class_ids[i] not in [2, 5, 7]:  # cars, buses, trucks
                    continue

                car_crop = frame[int(y1):int(y2), int(x1):int(x2)]
                if car_crop.shape[0] == 0 or car_crop.shape[1] == 0:
                    continue

                car_crop_resized = cv2.resize(car_crop, (640, 640)) / 255.0
                car_crop_tensor = torch.tensor(car_crop_resized).permute(2, 0, 1).unsqueeze(0).float()

                classification_results = classification_model(car_crop_tensor)
                pred_probs = classification_results[0].probs
                pred_index = torch.argmax(pred_probs.data).item()
                confidence = pred_probs.data[pred_index].item()

                if confidence < CLASSIFICATION_CONF_THRESH:
                    class_name = "Unknown"
                else:
                    class_name = classification_model.names[pred_index]
                    if class_name == "Suzuki_carry":
                        class_name = "Unknown"

                # Optionally save crops
                if save_crops and root_folder:
                    save_car_crop(car_crop, class_name, root_folder, class_counts, confidence)

                # Adding LPR
                # add_lpr_to_frame(frame, x1, y1, x2, y2, car_crop)

                # Draw car box and label
                cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 0, 255), 2)
                cv2.putText(
                    frame,
                    f"{class_name} ({confidence:.2f})",
                    (int(x1), max(int(y1) - 3, 0)),
                    cv2.FONT_HERSHEY_DUPLEX,  # smoother, more professional font
                    0.75,                     # same font scale
                    (0, 0, 255),              # red color
                    2,                        # thicker stroke
                    cv2.LINE_AA               # anti-aliasing
                  )

        out.write(frame)
        print(f"Frame {frame_count} processed in {time.time() - start_time:.2f} seconds")

    cap.release()
    out.release()
    print("Processing complete.")


# ---------------------- Helper to Save Car Crops ----------------------

def save_car_crop(display_crop, class_name, root_folder, class_counts, confidence):
    folder = os.path.join("/content", root_folder, class_name)
    os.makedirs(folder, exist_ok=True)
    count = class_counts[class_name]
    if class_name == "Unknown":
        filename = f"{class_name}_{count:03d}.jpg"
    else:
        filename = f"{class_name}_{count:03d}_conf{confidence:.2f}.jpg"
    save_path = os.path.join(folder, filename)
    class_counts[class_name] += 1
    cv2.imwrite(save_path, display_crop)
```

```python
from fast_plate_ocr import LicensePlateRecognizer
from transformers import pipeline
from PIL import Image
import re
```

```python
##################################
######### lpr_module.py ##########
##################################



license_plate_pattern = re.compile(
    r'(?i)[A-HJ-NP-TV-Z]{1,3}[ -]?\d{1,4}|\b[A-HJ-NP-TV-Z]{3}\b|\b\d{3,4}\b'
)

pipe = pipeline("object-detection", model="nickmuchi/yolos-small-finetuned-license-plate-detection")
ocr_model = LicensePlateRecognizer("cct-xs-v1-global-model")


# ---------------------- LPR Functions ----------------------

def detect_license_plate(car_crop):
    pil_img = Image.fromarray(cv2.cvtColor(car_crop, cv2.COLOR_BGR2RGB))
    results = pipe(pil_img)

    boxes = []
    for obj in results:
        if obj["score"] < 0.5:
            continue
        box = obj["box"]
        x1, y1, x2, y2 = int(box["xmin"]), int(box["ymin"]), int(box["xmax"]), int(box["ymax"])
        boxes.append((x1, y1, x2, y2))
    return boxes


def run_ocr_on_plate(plate_crop):
    text = ocr_model.run(plate_crop)
    return text[0] if text and text[0].strip() else "Unrecognizable"


def extract_valid_plate(text):
    match = license_plate_pattern.search(text)
    return match.group().strip() if match else "Unknown"


def draw_text_on_bbox(image, text, box, font_scale=0.6, color=(255, 255, 0)):
    x1, y1, x2, y2 = box
    label_pos = (x1, y1 - 10 if y1 - 10 > 10 else y1 + 20)
    cv2.putText(image, text, label_pos, cv2.FONT_HERSHEY_SIMPLEX, font_scale, color, 2, cv2.LINE_AA)


def add_lpr_to_frame(frame, x1, y1, x2, y2, car_crop):
    """
    Detect and annotate license plate for a given car crop on the main frame.
    """
    license_boxes = detect_license_plate(car_crop)
    plate_text = "Unknown"

    for (px1, py1, px2, py2) in license_boxes:
        plate_crop = car_crop[py1:py2, px1:px2]
        if plate_crop.shape[0] == 0 or plate_crop.shape[1] == 0:
            continue

        raw_text = run_ocr_on_plate(plate_crop)
        plate_text = extract_valid_plate(raw_text)

        global_px1, global_py1 = int(x1 + px1), int(y1 + py1)
        global_px2, global_py2 = int(x1 + px2), int(y1 + py2)

        cv2.rectangle(frame, (global_px1, global_py1), (global_px2, global_py2), (255, 0, 0), 2)

        # Plate text top-right of car box
        text_pos_x = int(x2) - 150
        text_pos_y = int(y1) + 20
        cv2.putText(frame, plate_text, (text_pos_x, text_pos_y),
                    cv2.FONT_HERSHEY_DUPLEX, 0.75, (0, 0, 255), 2, cv2.LINE_AA)
```

```python
#########################
#### utils_module.py ####
#########################

def adjust_gamma(image, gamma):
    invGamma = 1.0 / gamma
    table = np.array([(i / 255.0) ** invGamma * 255 for i in np.arange(256)]).astype("uint8")
    return cv2.LUT(image, table)
```

```python
########################################
#### night_classification_module.py ####
########################################


import os
import cv2
import time
import torch
from collections import defaultdict

# ---------------------- Car Detection + Classification ----------------------

def process_video(
    video_path,
    detection_model,
    classification_model,
    output_path,
    save_crops=False,
    root_folder=None,
    DETECTION_CONF_THRESH=0.0,
    CLASSIFICATION_CONF_THRESH=0.0,
):
    cap = cv2.VideoCapture(video_path)

    # Getting video properties
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Defining output video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    detection_model.eval()
    classification_model.eval()

    frame_count = 0
    class_counts = defaultdict(int)

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            print('End of video')
            break

        frame_count += 1
        print(f"Processing frame {frame_count}")
        start_time = time.time()

        detection_results = detection_model(frame)

        for result in detection_results:
            bboxes = result.boxes.xyxy.cpu().numpy()
            scores = result.boxes.conf.cpu().numpy()
            class_ids = result.boxes.cls.cpu().numpy()

            for i, (x1, y1, x2, y2) in enumerate(bboxes):
                if scores[i] < DETECTION_CONF_THRESH:
                    continue
                if class_ids[i] not in [2, 5, 7]:  # cars, buses, trucks
                    continue

                car_crop = frame[int(y1):int(y2), int(x1):int(x2)]
                if car_crop.shape[0] == 0 or car_crop.shape[1] == 0:
                    continue

                car_crop_gamma_corrected = adjust_gamma(car_crop, gamma=1.3)
                car_crop_resized = cv2.resize(car_crop_gamma_corrected, (640, 640)) / 255.0
                car_crop_tensor = torch.tensor(car_crop_resized).permute(2, 0, 1).unsqueeze(0).float()

                classification_results = classification_model(car_crop_tensor)
                pred_probs = classification_results[0].probs
                pred_index = torch.argmax(pred_probs.data).item()
                confidence = pred_probs.data[pred_index].item()

                if confidence < CLASSIFICATION_CONF_THRESH:
                    class_name = "Unknown"
                else:
                    class_name = classification_model.names[pred_index]
                    if class_name == "Suzuki_carry":
                        class_name = "Unknown"

                # Optionally save crops
                if save_crops and root_folder:
                    save_car_crop(car_crop, class_name, root_folder, class_counts, confidence)

                # Adding LPR
                add_lpr_to_frame(frame, x1, y1, x2, y2, car_crop)

                # Draw car box and label
                cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 0, 255), 2)
                cv2.putText(
                    frame,
                    f"{class_name} ({confidence:.2f})",
                    (int(x1), max(int(y1) - 3, 0)),
                    cv2.FONT_HERSHEY_DUPLEX,  # smoother, more professional font
                    0.75,                     # same font scale
                    (0, 0, 255),              # red color
                    2,                        # thicker stroke
                    cv2.LINE_AA               # anti-aliasing
                  )

        out.write(frame)
        print(f"Frame {frame_count} processed in {time.time() - start_time:.2f} seconds")

    cap.release()
    out.release()
    print("Processing complete.")


# ---------------------- Helper to Save Car Crops ----------------------

def save_car_crop(display_crop, class_name, root_folder, class_counts, confidence):
    folder = os.path.join("/content", root_folder, class_name)
    os.makedirs(folder, exist_ok=True)
    count = class_counts[class_name]
    if class_name == "Unknown":
        filename = f"{class_name}_{count:03d}.jpg"
    else:
        filename = f"{class_name}_{count:03d}_conf{confidence:.2f}.jpg"
    save_path = os.path.join(folder, filename)
    class_counts[class_name] += 1
    cv2.imwrite(save_path, display_crop)
```

```python
from fast_plate_ocr import LicensePlateRecognizer
from transformers import pipeline
from PIL import Image
import re
```

```python
########################################
######### night_lpr_module.py ##########
########################################



license_plate_pattern = re.compile(
    r'(?i)[A-HJ-NP-TV-Z]{1,3}[ -]?\d{1,4}|\b[A-HJ-NP-TV-Z]{3}\b|\b\d{3,4}\b'
)

pipe = pipeline("object-detection", model="nickmuchi/yolos-small-finetuned-license-plate-detection")
ocr_model = LicensePlateRecognizer("cct-xs-v1-global-model")


# ---------------------- LPR Functions ----------------------

def detect_license_plate(car_crop):
    pil_img = Image.fromarray(cv2.cvtColor(car_crop, cv2.COLOR_BGR2RGB))
    results = pipe(pil_img)

    boxes = []
    for obj in results:
        if obj["score"] < 0.5:
            continue
        box = obj["box"]
        x1, y1, x2, y2 = int(box["xmin"]), int(box["ymin"]), int(box["xmax"]), int(box["ymax"])
        boxes.append((x1, y1, x2, y2))
    return boxes


def run_ocr_on_plate(plate_crop):
    gray = cv2.cvtColor(plate_crop, cv2.COLOR_BGR2GRAY)
    invGamma = 1.0 / 1.5
    table = np.array([(i / 255.0) ** invGamma * 255 for i in np.arange(256)]).astype("uint8")
    gamma_corrected = cv2.LUT(gray, table)
    clahe = cv2.createCLAHE(clipLimit=20.0, tileGridSize=(8, 8))
    clahe_image = clahe.apply(gamma_corrected)
    enhanced_plate = cv2.cvtColor(clahe_image, cv2.COLOR_GRAY2BGR)

    text = ocr_model.run(enhanced_plate)
    return text[0] if text and text[0].strip() else "Unrecognizable"


def extract_valid_plate(text):
    match = license_plate_pattern.search(text)
    return match.group().strip() if match else "Unknown"


def draw_text_on_bbox(image, text, box, font_scale=0.6, color=(255, 255, 0)):
    x1, y1, x2, y2 = box
    label_pos = (x1, y1 - 10 if y1 - 10 > 10 else y1 + 20)
    cv2.putText(image, text, label_pos, cv2.FONT_HERSHEY_SIMPLEX, font_scale, color, 2, cv2.LINE_AA)


def add_lpr_to_frame(frame, x1, y1, x2, y2, car_crop):
    """
    Detect and annotate license plate for a given car crop on the main frame.
    """
    license_boxes = detect_license_plate(car_crop)
    plate_text = "Unknown"

    for (px1, py1, px2, py2) in license_boxes:
        plate_crop = car_crop[py1:py2, px1:px2]
        if plate_crop.shape[0] == 0 or plate_crop.shape[1] == 0:
            continue

        raw_text = run_ocr_on_plate(plate_crop)
        plate_text = extract_valid_plate(raw_text)

        global_px1, global_py1 = int(x1 + px1), int(y1 + py1)
        global_px2, global_py2 = int(x1 + px2), int(y1 + py2)

        cv2.rectangle(frame, (global_px1, global_py1), (global_px2, global_py2), (255, 0, 0), 2)

        # Plate text top-right of car box
        text_pos_x = int(x2) - 150
        text_pos_y = int(y1) + 20
        cv2.putText(frame, plate_text, (text_pos_x, text_pos_y),
                    cv2.FONT_HERSHEY_DUPLEX, 0.75, (0, 0, 255), 2, cv2.LINE_AA)
```
